```python
import statistics as sts
import seaborn as srn
import pandas as pd
```

```python
#importando o arquivo csv que iremos tratar
dataset = pd.read_csv("tempo.csv", sep = ";")

dataset.head()
```

```python
#verificando o shape
dataset.shape
```

```python
#verificando valores da coluna Aparência
agrupadoAp = dataset.groupby(['Aparencia']).size()
agrupadoAp
```

```python
#verificando a existência de valores NAs na coluna Aparência
dataset["Aparencia"].isnull().sum()
```

```python
#tratando a coluna Aparência
dataset.loc[dataset["Aparencia"] == "menos", "Aparencia"] = "sol"
#verificando o resultado
agrupadoAp = dataset.groupby(['Aparencia']).size()
agrupadoAp
```

```python
#verificando o range de valores presentes na coluna Temperatura
dataset["Temperatura"].describe()
```

```python
#verificando existência de valores NAs
dataset["Temperatura"].isnull().sum()
```

```python
#tratando a coluna Temperatura (substituindo valores < -130 e > 130 pela mediana)
medianaTemp = sts.median(dataset["Temperatura"])
dataset.loc[(dataset["Temperatura"] < -130) | (dataset["Temperatura"] > 130), "Temperatura"] = medianaTemp
#verificando o resultado
dataset["Temperatura"].describe()
```

```python
#verificando o range de valores na coluna Umidade
dataset["Umidade"].describe()
```

```python
#verificando existência de valores NAs
dataset["Umidade"].isnull().sum()
```

```python
#tratando a coluna Umidade (substituindo valores < 0 e > 100 e valores NAs pela mediana)
medianaUmi = sts.median(dataset["Umidade"])
dataset.loc[(dataset["Umidade"] < 0) | (dataset["Umidade"] > 100), "Umidade"] = medianaUmi
dataset["Umidade"].fillna(medianaUmi, inplace = True)
#verificando o resultado
dataset["Umidade"].describe()
```

```python
#verificando valores da coluna Vento
agrupadoVento = dataset.groupby(["Vento"]).size()
agrupadoVento
```

```python
#verificando a existência de valores NAs na coluna Vento
dataset["Vento"].isnull().sum()
```

```python
#tratando a coluna Vento
dataset["Vento"].fillna("FALSO", inplace = True)
#verificando o resultado
agrupadoAp = dataset.groupby(["Vento"]).size()
agrupadoAp
```
